```python
import plotly.graph_objects as go
import pandas as pd

data = [
    # id, parent, label, desc, value
    ["root", "", "CIR-IF Taxonomy", "", 0],
    ["human", "root", "Human Factors", "", 0],
    ["context", "root", "Context Factors", "", 0],
    ["knowledge", "human", "Knowledge", "", 0],
    ["cptb", "human", "CPTB", "", 0],
    ["org", "context", "Organization", "", 0],
    
    # Leaves with descriptions and values
    ["skills", "knowledge", "Skills & Hard Knowledge", "Training and tacit domain expertise", 23],
    ["sa", "knowledge", "Situation Awareness", "Adversary TTP recognition and assessment", 23],
    ["stress", "cptb", "Stress Handling", "Fatigue, cognitive load, burnout management", 23],
    ["mgmt", "org", "Management", "Leadership metrics, workload, and HR support", 19],
]

df = pd.DataFrame(data, columns=["id", "parent", "label", "desc", "val"])

fig = go.Figure(go.Icicle(
    ids=df["id"],
    labels=df["label"],
    parents=df["parent"],
    values=df["val"],
    branchvalues="remainder",
    customdata=df[["desc"]].values,
    texttemplate="<b>%{label}</b><br><span style='font-size:10px;'>%{customdata[0]}</span>",
    textposition="middle center"
))

fig.update_layout(margin=dict(t=20, l=10, r=10, b=10))
fig.show()
```

```python
# Example using CIR-IF low-level subcategories
data = {
    "ids": ["Human", "Knowledge", "Skills", "SA"],
    "parents": ["", "Human", "Knowledge", "Knowledge"],
    "labels": ["Human Factors", "Knowledge", "Knowledge & Skills", "Situation Awareness"],
    "values": [39, 22, 13, 9],
    "descriptions": [
        "Root human domain",
        "Domain knowledge category",
        "Hard/soft skills, training, and tacit expertise across defenders",
        "Perception, comprehension, and projection of adversary TTPs"
    ]
}

df = pd.DataFrame(data)

fig = px.icicle(
    df,
    ids="ids",
    names="labels",
    parents="parents",
    values="values",
    custom_data=["descriptions"]
)

# %{customdata[0]} pulls in the corresponding description text
fig.update_traces(
    hovertemplate="<b>%{label}</b><br>Value: %{value}<br><br><i>%{customdata[0]}</i><extra></extra>"
)

fig.show()
```
